```python
from matplotlib import pyplot as plt
import numpy as np
import random


# Creating dataset
a = []

for i in range(80000):
  v1 = random.uniform(0.5, 1.1)
  v2 = random.uniform(0.45, 1)
  a.append(max(v1, v2))
  # a.append(random.uniform(0.5, 1))

# Creating histogram
fig, ax = plt.subplots(figsize =(10, 7))
ax.hist(a)

# Show plot
plt.show()
```

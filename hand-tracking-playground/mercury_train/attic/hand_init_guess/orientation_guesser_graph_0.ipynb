```python
c = pd.read_csv("/4/clones/mercury_train/attic/guess_vs_actual_1_inv.csv"); #, delimiter = " ")
n = np.array(c)


fig, ax = plt.subplots(7, 1, figsize=(40, 40))


def quat_diff(one, two):
    inner_product = (one[:,0]*two[:,0]) + (one[:,1]*two[:,1]) + (one[:,2]*two[:,2]) + (one[:,3]*two[:,3])
    print(np.ones(one.shape).shape)

    return np.ones([one.shape[0]])-(inner_product*inner_product)


print(n[:, 10:14].shape)
print(n[:, 3:7].shape)

ax[0].plot(quat_diff(n[:, 3:7], n[:, 10:14]))

ax[0].set_ylim([-1, 1])
ax[1].set_ylim([-1, 1])
ax[2].set_ylim([-1, 1])

ax[3].set_ylim([-1, 1])
ax[4].set_ylim([-1, 1])
ax[5].set_ylim([-1, 1])
ax[6].set_ylim([-1, 1])


# ax.plot(c[1], color="green")
# ax.plot(c[2], color="blue")""
```

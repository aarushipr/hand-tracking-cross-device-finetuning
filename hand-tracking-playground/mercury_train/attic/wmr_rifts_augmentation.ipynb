```python
import random
import numpy as np
import cv2
import matplotlib.pyplot as plt

def maybe_too_much_contrast_moshi(input_: np.ndarray):


    img_min = int(input_.min())-5
    img_max = int(input_.max())+5

    one = int(random.randint(img_min, img_max))



    two = int(random.randint(img_min, img_max))




    the_min = min(one, two)
    the_max = max(one, two)

    if ((the_max - the_min) < 1):
        # /shrug
        print("Increasing")
        the_max += 1
        the_min += 1


    input_ = np.clip(input_, int(the_min), int(the_max))

    if np.std(input_) < 0.0000001:
        print("Uh oh!", one, two)

    
    return input_

def maybe_too_dark_or_too_low_contrast(img_: np.ndarray):
    img_ = img_.astype(np.float32)
    scale_amt = np.random.uniform(2/255, 50/255)
    # scale_amt = 3/255

    img_ *= scale_amt
    img_ = img_.astype(np.uint8)
    return img_


def mat_float32touint8(mat):
    return (np.clip(mat, 0, 1) * 255).astype(np.uint8)


def mat_uint8tofloat32(mat):
    return mat.astype(np.float32) * 1.0 / 255.0




def normalizeGrayscaleImage(
        img,
        report=None,
        target_mean=0.5,
        target_std=0.25):
    if img.dtype == np.uint8:
        img = mat_uint8tofloat32(img)
    std = np.std(img)
    if np.allclose(img, img[0, 0]):
        print(f"Very low contrast: {report}")
        # exit()
        # img = np.random.random(img.shape)
        # std = np.std(img)
    img *= target_std / std
    img += target_mean - np.mean(img)

    np.clip(img, 0, 1)

    return img


def add_noise(img):
    return img + np.random.normal(0, np.random.uniform(2, 15), img.shape).astype(np.uint8)








# %matplotlib inline
img_orig = cv2.imread("/4/datasets/inshallah10/seq1061/imgs_color/Image0003.jpg", cv2.IMREAD_GRAYSCALE)

normalize = True
if normalize:
    max_display = 1
else:
    max_display = 255


fig, ax = plt.subplots(3, 10, figsize=(50, 15))

if normalize:
    ax[0, 0].imshow(normalizeGrayscaleImage(img_orig), cmap='gray', vmin=0, vmax=1)
else:
    ax[0, 0].imshow(img_orig, cmap='gray', vmin=0, vmax=max_display)

for i in range(1, 10):
    ax[0, i].imshow(np.zeros((128, 128)))

for i in range(10):
    img = maybe_too_much_contrast_moshi(img_orig)
    img = add_noise(img)

    if normalize:
        img = normalizeGrayscaleImage(img)


    ax[1, i].imshow(img, cmap='gray', vmin=0, vmax=max_display)

    # print(img.min(), img.max())

for i in range(10):
    img = maybe_too_dark_or_too_low_contrast(img_orig)
    img = add_noise(img)
    # img += np.random.normal(0, np.random.uniform(2, 25), img.shape).astype(np.uint8)

    if normalize:
        img = normalizeGrayscaleImage(img)



    ax[2, i].imshow(img, cmap='gray', vmin=0, vmax=max_display)

    # print(img.min(), img.max())

```

```python
import sympy
from sympy import symbols
from sympy.utilities.codegen import codegen


def normalize(x, y, z):
    len = sympy.sqrt(x*x + y*y + z*z)
    return (x/len, y/len, z/len)


def sg(x, y, z):
    # denom = 1/(1-z)
    return (x/(1-z), y/(1-z))

def map_ranges(value, from_low, from_high, to_low, to_high):
	return (value - from_low) * (to_high - to_low) / (from_high - from_low) + to_low;

# Old version. Not clear why I thought it was important to map this to pixel-coords.
def img_from_sg(u, v, ra):
    u_ = map_ranges(u, -ra, ra, 0, 128);
    v_ = map_ranges(v, ra, -ra, 0, 128);
    return (u_, v_)

# This one feels right
def img_from_sg(u, v, ra):
    u_ = map_ranges(u, -ra, ra, -1, 1)
    v_ = map_ranges(v, ra, -ra, -1, 1)
    return (u_, v_)




x, y, z, u, v, u0, v0, sr = symbols('x y z u v u0 v0 sr')

# u0, v0 = sg(*normalize(x, y, z))

u0, v0 = sg(x,y,z)

u, v = img_from_sg(u0, v0, sr)

print("float du_dx =", sympy.ccode(sympy.simplify(u.diff(x))), ";")
print("float du_dy =", sympy.ccode(sympy.simplify(u.diff(y))), ";")
print("float du_dz =", sympy.ccode(sympy.simplify(u.diff(z))), ";")
print()
print("float dv_dx =", sympy.ccode(sympy.simplify(v.diff(x))), ";")
print("float dv_dy =", sympy.ccode(sympy.simplify(v.diff(y))), ";")
print("float dv_dz =", sympy.ccode(sympy.simplify(v.diff(z))), ";")

print()
print()

print("float du_dx =", sympy.ccode(sympy.simplify(sympy.diff(u, x)), type_aliases={sympy.codegen.ast.real: sympy.codegen.ast.float32}), ";")
print("float du_dy =", sympy.ccode(sympy.simplify(sympy.diff(u, y)), type_aliases={sympy.codegen.ast.real: sympy.codegen.ast.float32}), ";")
print("float du_dz =", sympy.ccode(sympy.simplify(sympy.diff(u, z)), type_aliases={sympy.codegen.ast.real: sympy.codegen.ast.float32}), ";")
print()
print("float dv_dx =", sympy.ccode(sympy.simplify(sympy.diff(v, x)), type_aliases={sympy.codegen.ast.real: sympy.codegen.ast.float32}), ";")
print("float dv_dy =", sympy.ccode(sympy.simplify(sympy.diff(v, y)), type_aliases={sympy.codegen.ast.real: sympy.codegen.ast.float32}), ";")
print("float dv_dz =", sympy.ccode(sympy.simplify(sympy.diff(v, z)), type_aliases={sympy.codegen.ast.real: sympy.codegen.ast.float32}), ";")


```

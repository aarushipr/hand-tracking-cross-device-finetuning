```python
# Move me back to `py_training` if you need me!
import torch
import torch.nn as nn
import os

import KeyNet
import KeyNet_Old
model = KeyNet_Old.KeyNet()
model_new = KeyNet.KeyNet()

def save_checkpoint(states, output_dir, filename='checkpoint.pth'):
    os.makedirs(output_dir, exist_ok=True)
    torch.save(states, os.path.join(output_dir, filename))


checkpoint_file = os.path.join(
    "checkpoints", 'checkpoint.pth'
)

checkpoint_file = "/4/clones/artificial_data_generator/py_training/checkpoints_jan11_predicted/checkpoint.pth"
checkpoint_file_new = "/4/clones/artificial_data_generator/py_training/checkpoints/checkpoint.pth"

checkpoint = torch.load(
    checkpoint_file, map_location=torch.device('cpu'))

print(checkpoint["state_dict"].keys())
    
model.load_state_dict(checkpoint["state_dict"])
# print("loaded state dict")

model_new.image_network.load_state_dict(model.image_network.state_dict())
# model_new.keypoints_network.load_state_dict(model.keypoints_network.state_dict()) # Not used; different shape
model_new.fused_network.load_state_dict(model.fused_network.state_dict())
model_new.network_1d_depth.load_state_dict(model.network_1d_depth.state_dict())
model_new.network_extras.load_state_dict(model.network_extras.state_dict())
model_new.network_curls.load_state_dict(model.network_curls.state_dict())
model_new.network_2d_px_coord.load_state_dict(model.network_2d_px_coord.state_dict())

save_checkpoint({
    'epoch': checkpoint['epoch'],
    'state_dict': model_new.state_dict(),
}, "checkpoints")


```
